## 4. Remove Duplicates

```python
# Training data - remove duplicates
benf_before = len(benf)
inpatient_before = len(inpatient)
outpatient_before = len(outpatient)

benf = benf.drop_duplicates()
inpatient = inpatient.drop_duplicates()
outpatient = outpatient.drop_duplicates()

print("Training Data Duplicates Removed:")
print(f"  Beneficiary: {benf_before} -> {len(benf)} (removed {benf_before - len(benf)})")
print(f"  Inpatient: {inpatient_before} -> {len(inpatient)} (removed {inpatient_before - len(inpatient)})")
print(f"  Outpatient: {outpatient_before} -> {len(outpatient)} (removed {outpatient_before - len(outpatient)})")
```

```python
# Test data - remove duplicates
benf_test_before = len(benf_test)
inpatient_test_before = len(inpatient_test)
outpatient_test_before = len(outpatient_test)

benf_test = benf_test.drop_duplicates()
inpatient_test = inpatient_test.drop_duplicates()
outpatient_test = outpatient_test.drop_duplicates()

print("Test Data Duplicates Removed:")
print(f"  Beneficiary: {benf_test_before} -> {len(benf_test)} (removed {benf_test_before - len(benf_test)})")
print(f"  Inpatient: {inpatient_test_before} -> {len(inpatient_test)} (removed {inpatient_test_before - len(inpatient_test)})")
print(f"  Outpatient: {outpatient_test_before} -> {len(outpatient_test)} (removed {outpatient_test_before - len(outpatient_test)})")
```

## 5. Data Quality Analysis

### 5.1 Missing Values Report

```python
def missing_report(df, name):
    """Generate a report of missing values in a dataframe"""
    print(f"\n{'='*60}")
    print(f"Missing Value Report – {name}")
    print(f"{'='*60}")
    missing_pct = df.isnull().mean().sort_values(ascending=False).head(20)
    if len(missing_pct) > 0:
        print(missing_pct)
    else:
        print("No missing values found!")

missing_report(benf, "Beneficiary (Training)")
missing_report(inpatient, "Inpatient (Training)")
missing_report(outpatient, "Outpatient (Training)")
```
