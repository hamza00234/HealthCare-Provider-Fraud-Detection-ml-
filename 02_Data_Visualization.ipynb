# Data Visualization
## Healthcare Provider Fraud Detection

This notebook focuses on visualizing data patterns, distributions, and relationships.

## 1. Setup and Data Loading

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

```python
# Load data from Data Loading notebook or directly
benf = pd.read_csv('Train_Beneficiarydata-1542865627584.csv').drop_duplicates()
inpatient = pd.read_csv('Train_Inpatientdata-1542865627584.csv').drop_duplicates()
outpatient = pd.read_csv('Train_Outpatientdata-1542865627584.csv').drop_duplicates()
train_labels = pd.read_csv('Train-1542865627584.csv')

print("Data loaded successfully!")
```

## 2. Missing Values Visualization

### 2.1 Inpatient Missing Data Heatmap

```python
missing_cols_in = inpatient.columns[inpatient.isnull().any()]
plt.figure(figsize=(14, 8))
sns.heatmap(inpatient[missing_cols_in].isnull(), cbar=True, cmap='YlOrRd')
plt.title("Missingness Map – Inpatient (columns with missing data)", fontsize=14, fontweight='bold')
plt.xlabel("Columns", fontsize=12)
plt.ylabel("Records", fontsize=12)
plt.tight_layout()
plt.show()
```

### 2.2 Outpatient Missing Data Heatmap

```python
missing_cols_out = outpatient.columns[outpatient.isnull().any()]
plt.figure(figsize=(14, 8))
sns.heatmap(outpatient[missing_cols_out].isnull(), cbar=True, cmap='YlOrRd')
plt.title("Missingness Map – Outpatient (columns with missing data)", fontsize=14, fontweight='bold')
plt.xlabel("Columns", fontsize=12)
plt.ylabel("Records", fontsize=12)
plt.tight_layout()
plt.show()
```

### 2.3 Missing Values Percentage Bar Charts

```python
# Inpatient Missing Percentages
in_missing_pct = inpatient.isnull().mean() * 100
in_missing_pct = in_missing_pct[in_missing_pct > 0].sort_values(ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=in_missing_pct.index, y=in_missing_pct.values, palette='viridis')
plt.xticks(rotation=45, ha='right')
plt.ylabel("Percentage Missing (%)", fontsize=12)
plt.title("Percentage of Missing Values per Column – Inpatient", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print(f"\nInpatient Missing Values (%):\n{in_missing_pct}")
```

```python
# Outpatient Missing Percentages
out_missing_pct = outpatient.isnull().mean() * 100
out_missing_pct = out_missing_pct[out_missing_pct > 0].sort_values(ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=out_missing_pct.index, y=out_missing_pct.values, palette='plasma')
plt.xticks(rotation=45, ha='right')
plt.ylabel("Percentage Missing (%)", fontsize=12)
plt.title("Percentage of Missing Values per Column – Outpatient", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print(f"\nOutpatient Missing Values (%):\n{out_missing_pct}")
```

## 3. Target Variable Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Count plot
fraud_counts = train_labels['PotentialFraud'].value_counts()
sns.barplot(x=fraud_counts.index, y=fraud_counts.values, ax=axes[0], palette='Set2')
axes[0].set_title("Fraud Distribution - Count", fontsize=12, fontweight='bold')
axes[0].set_xlabel("Fraud Label")
axes[0].set_ylabel("Count")
for i, v in enumerate(fraud_counts.values):
    axes[0].text(i, v + 2, str(v), ha='center', fontweight='bold')

# Pie chart
fraud_pct = train_labels['PotentialFraud'].value_counts() / len(train_labels) * 100
colors = ['#66c2a5', '#fc8d62']
axes[1].pie(fraud_pct.values, labels=['Non-Fraud', 'Fraud'], autopct='%1.1f%%', 
           colors=colors, startangle=90, textprops={'fontsize': 11, 'fontweight': 'bold'})
axes[1].set_title("Fraud Distribution - Percentage", fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"\nTarget Variable Distribution:\n{fraud_counts}")
```

## 4. Numerical Features Distribution

```python
# Select numerical columns from beneficiary data
numerical_cols = benf.select_dtypes(include=[np.number]).columns.tolist()

fig, axes = plt.subplots(len(numerical_cols), 2, figsize=(16, 4*len(numerical_cols)))

for idx, col in enumerate(numerical_cols):
    # Histogram
    axes[idx, 0].hist(benf[col].dropna(), bins=50, color='skyblue', edgecolor='black', alpha=0.7)
    axes[idx, 0].set_title(f"{col} - Distribution", fontweight='bold')
    axes[idx, 0].set_xlabel(col)
    axes[idx, 0].set_ylabel("Frequency")
    axes[idx, 0].grid(alpha=0.3)
    
    # Box plot
    axes[idx, 1].boxplot(benf[col].dropna(), vert=True)
    axes[idx, 1].set_title(f"{col} - Box Plot", fontweight='bold')
    axes[idx, 1].set_ylabel(col)
    axes[idx, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Categorical Features Distribution

```python
# Select categorical columns from beneficiary data
categorical_cols = benf.select_dtypes(include=['object']).columns.tolist()

for col in categorical_cols:
    value_counts = benf[col].value_counts()
    if len(value_counts) <= 10:
        plt.figure(figsize=(10, 5))
        sns.barplot(x=value_counts.index, y=value_counts.values, palette='Set2')
        plt.title(f"{col} - Distribution", fontsize=12, fontweight='bold')
        plt.ylabel("Count")
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

## 6. Correlation Analysis

```python
# Beneficiary numerical correlations
numeric_benf = benf.select_dtypes(include=[np.number])
corr_matrix = numeric_benf.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            cbar_kws={'label': 'Correlation'}, square=True)
plt.title("Correlation Matrix - Beneficiary Data", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
